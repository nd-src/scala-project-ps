```scala
class Employee {var salary = 100}
```

```scala
val John = new Employee
```

```scala
John.salary
```

```scala
John.salary = 5000
```

```scala
class Employee {
    private var salary: Int = 0

    def getSalary() = salary

    def setSalary(newSalary: Int) = {
        salary = newSalary
    }
}
```

```scala
val Johnny = new Employee
```

```scala
Johnny.salary
```

```scala
Johnny.getSalary
```

```scala
Johnny.getSalary = 5000
```

```scala
Johnny.setSalary(5000)
```

```scala
Johnny.getSalary
```
