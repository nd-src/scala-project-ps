## If statement in Scala is an expression - It YIELDS a value

````scala
val argument = Array("Monday")

val day = if (!argument.isEmpty) {
    argument(0);
} else {
    "Sunday"
}
````

## The LAST statement **becomes** the RETURN value

````scala
// one line code dont need curly braces
val someDay = if (!argument.isEmpty) argument(0) else "Sunday"
````

## Loops: for loops: Imperative loops vs Functional loops (returns a value)

````scala
val numbers = List(1,2,3,4,5,6,8,9,10,11)

// imperative loops
for (number <- numbers) {
    if (number % 2 == 0) {
        println(number)
    }
}
````

````scala
// functional loop
for (number <- numbers if (number % 2 == 0)) println(number)
````

````scala
for (number <- numbers if (number % 2 == 0)) {
    println(number)
}
````

````scala
for (number <- numbers if (number % 2 == 0))
    println(number)
````

Multiple if conditions

````scala
for (
    number <- numbers
    if (number % 2 == 0)
    if (number > 4)
) println(number)
````

````scala
// this fails

for number <- numbers
    if (number % 2 == 0)
    if (number > 4)
        println(number)
````

````scala
// this fails
// this could work in Scala 3
for number <- numbers
    if (number % 2 == 0)
    if (number > 4)
do 
    println(number)
````

### Double for loops - IMPERATIVE

````scala
var newNums = List(1,2)
var newLetters = List('a','b','c')

for (number <- newNums) {
    for (letter <- newLetters) {
        println(number, letter)
    }
}
````

````scala
// another way to write the same thing
for {
    number <- newNums
    letter <- newLetters
} println(number, letter)
````

````scala
// another way to write the same thing BUT WITH BRACKETS
// new lines inside a bracket need ; in Scala 2
for (
    number <- newNums;
    letter <- newLetters
) println(number, letter)
````

```scala
// in Scala 3, the following would also work
for
  x <- xs
  y <- ys
do
  println(x, y)
```

### Double for loops - FUNCTIONAL

```scala

for clause yield body

````scala
var squaredNewNums = for (number <- newNums) yield number * 2
squaredNewNums
````

````scala
// newNums stays the same
newNums
````

````scala
// multiple yield - type collision
squaredNewNums = for (number <- newNums) yield {
    println(number, number * 2)   // this returns List[Unit] type
}
squaredNewNums
````

````scala
// multiple yield
squaredNewNums = for (number <- newNums) yield {
    number * 2          // this value is discarded
    println(number *3)  // this is printed
    number + 100        // only this is returned
}
squaredNewNums
````

FUNCTIONAL for loop with if

````scala
for (number <- newNums if (number > 1)) yield number
````

````scala
newNums
````

````scala
for (
    number <- newNums;    // semi-colon required
    letter <- newLetters
) yield number + "->" + letter
````

````scala
for {
    number <- newNums    // semi-colon NOT required
    letter <- newLetters
} yield number + "->" + letter
````
