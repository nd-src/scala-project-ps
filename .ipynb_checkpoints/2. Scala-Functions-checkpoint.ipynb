## Functions

```scala

// return type is optional in Scala
def functionName(argument: Type): returnType = {body}

```

````scala
def add(a:Int, b:Int): Int = {a+b}
````

````scala
add(1,2)
````

````scala
add(1, "5")
````

````scala
def plusOne(number: Int) = {
    if (number < 0) 1000
    else number + 1
}
````

````scala
plusOne(-1)
````

````scala
plusOne(1)
````

````scala
def plusTwo(number: Int) = {
    if (number <= 0) "Must be greater than 0"
    else number + 2
}
````

````scala
plusTwo(-22)
````

````scala
plusTwo(2)
````

````scala
def product(a: Int, b: Int) = a * b
````
